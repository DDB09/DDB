```python
import py4hw

class DetectorSequence011(py4hw.Logic):
    def __init__(self, parent, name, reset, a, r):
        super().__init__(parent, name)
        self.reset = self.addIn('reset', reset)
        self.a = self.addIn('a', a)
        self.r = self.addOut('r', r)
        self.state = 0  # idle

    def clock(self):
        if (self.reset.get()):
            self.state = 0  # -> idle
            self.r.prepare(0)
        else:
            if (self.state == 0):  # Idle
                self.r.prepare(0)
                if (self.a.get() == 0):
                    self.state = 1  # -> 0 sequence

            elif (self.state == 1):  # 0 sequence
                self.r.prepare(0)
                if (self.a.get() == 1):
                    self.state = 2  # -> 01 sequence

            elif (self.state == 2):  # 01 sequence
                if (self.a.get() == 1):
                    self.r.prepare(1)
                    self.state = 3  # -> 011 sequence
                else:
                    self.r.prepare(0)
                    self.state = 1  # -> 0 sequence

            elif (self.state == 3):  # 011 sequence
                self.r.prepare(0)
                if (self.a.get() == 1):
                    self.state = 0  # -> idle
                else:
                    self.state = 1  # -> 0 sequence
```

```python
import py4hw

# Create a circuit
hw = py4hw.HWSystem()

# Create signals
reset = py4hw.Wire(hw, "reset")
a = py4hw.Wire(hw, "a")
r = py4hw.Wire(hw, "r")

# Create the detector
detector = DetectorSequence011(hw, "Detector011", reset, a, r)
```
